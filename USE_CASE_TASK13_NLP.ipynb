```python
from textblob import TextBlob

text = input("Enter a sentence: ")

analysis = TextBlob(text)
polarity = analysis.sentiment.polarity

if polarity > 0:
    emotion = "Positive"
elif polarity < 0:
    emotion = "Negative"
else:
    emotion = "Neutral"

print("Detected Emotion:", emotion)
```
